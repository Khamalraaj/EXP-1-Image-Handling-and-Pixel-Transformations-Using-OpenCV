14. Read the image ('boy.jpg').

```python
img = cv2.imread("boy.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

15. Adjust the brightness of the image.

```python
matrix = np.ones(img.shape, dtype='float64') * 50
```

16. Create brighter and darker images.

```python
img_brighter = cv2.add(img, matrix.astype(np.uint8))
img_darker = cv2.subtract(img, matrix.astype(np.uint8))
```

17. Display the images (Original Image, Darker Image, Brighter Image).

```python
plt.figure(figsize=(10,4))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title("Original")

plt.subplot(1,3,2)
plt.imshow(img_darker)
plt.title("Darker")

plt.subplot(1,3,3)
plt.imshow(img_brighter)
plt.title("Brighter")

plt.show()
```

18. Modify the image contrast.

```python
img_higher1 = cv2.convertScaleAbs(img, alpha=1.1, beta=0)
img_higher2 = cv2.convertScaleAbs(img, alpha=1.2, beta=0)
```

19. Display the images (Original, Lower Contrast, Higher Contrast).

```python
plt.figure(figsize=(10,4))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title("Original")

plt.subplot(1,3,2)
plt.imshow(img_higher1)
plt.title("Contrast 1.1")

plt.subplot(1,3,3)
plt.imshow(img_higher2)
plt.title("Contrast 1.2")

plt.show()
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
img_bgr2 = cv2.imread("boy.jpg")
b, g, r = cv2.split(img_bgr2)

plt.imshow(b, cmap='gray')
plt.title("Blue Channel")
plt.show()

plt.imshow(g, cmap='gray')
plt.title("Green Channel")
plt.show()

plt.imshow(r, cmap='gray')
plt.title("Red Channel")
plt.show()
```

21. _merged the R, G, B , display along with orginal image 

```python
merged = cv2.merge((b, g, r))
merged = cv2.cvtColor(merged, cv2.COLOR_BGR2RGB)

plt.imshow(merged)
plt.title("Merged Image")
plt.axis('off')
plt.show()
```

22. Split the image into the H, S, V components &  Display the channels.

```python
hsv = cv2.cvtColor(img_bgr2, cv2.COLOR_BGR2HSV)
h, s, v = cv2.split(hsv)

plt.imshow(h, cmap='gray')
plt.title("Hue")
plt.show()

plt.imshow(s, cmap='gray')
plt.title("Saturation")
plt.show()

plt.imshow(v, cmap='gray')
plt.title("Value")
plt.show()
```

23. _merged the H, S, V, display along with orginal image 

```python
merged_hsv = cv2.merge((h, s, v))
final_img = cv2.cvtColor(merged_hsv, cv2.COLOR_HSV2RGB)

plt.imshow(final_img)
plt.title("HSV Merged Image")
plt.axis('off')
plt.show()
```
